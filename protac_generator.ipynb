```python
import rdkit
from rdkit import Chem
from rdkit.Chem import Descriptors, QED, Draw
import random
import pandas as pd
from google.colab import files

# Step A: Define components with explicit attachment points (*)
thalidomide_anchor = "O=C1NC(=O)CCC1N2C(=O)c3ccc(*)cc3C2=O"
crtap_binder = "*CC(C)CC(C(=O)O)NC(=O)C"

print("AI Pipeline Initialized. Starting Linker Generation...")

# Step B: The AI Generator Engine
def generate_random_linker(length):
    atoms = ["C", "C", "O", "C", "C"]
    linker = "".join(random.choice(atoms) for _ in range(length))
    return f"*{linker}*"

# Step C: The Screening Filter
generated_protacs = []
seen_smiles = set()

mol_anchor = Chem.MolFromSmiles(thalidomide_anchor)
mol_binder = Chem.MolFromSmiles(crtap_binder)

if mol_anchor is not None and mol_binder is not None:
    for i in range(500):
        linker_length = random.randint(4, 12)
        linker_smiles = generate_random_linker(linker_length)
        mol_linker = Chem.MolFromSmiles(linker_smiles)

        if mol_linker is None:
            continue

        combined_1 = Chem.CombineMols(mol_anchor, mol_linker)
        editable_1 = Chem.EditableMol(combined_1)
        matches_1 = [atom.GetIdx() for atom in combined_1.GetAtoms() if atom.GetSymbol() == '*']

        if len(matches_1) >= 2:
            editable_1.AddBond(matches_1[0], matches_1[1], order=Chem.BondType.SINGLE)
            editable_1.RemoveAtom(max(matches_1[0], matches_1[1]))
            editable_1.RemoveAtom(min(matches_1[0], matches_1[1]))
            intermediate = editable_1.GetMol()

            combined_2 = Chem.CombineMols(intermediate, mol_binder)
            editable_2 = Chem.EditableMol(combined_2)
            matches_2 = [atom.GetIdx() for atom in combined_2.GetAtoms() if atom.GetSymbol() == '*']

            if len(matches_2) >= 2:
                editable_2.AddBond(matches_2[0], matches_2[1], order=Chem.BondType.SINGLE)
                editable_2.RemoveAtom(max(matches_2[0], matches_2[1]))
                editable_2.RemoveAtom(min(matches_2[0], matches_2[1]))
                final_mol = editable_2.GetMol()

                try:
                    Chem.SanitizeMol(final_mol)
                    canonical_smiles = Chem.MolToSmiles(final_mol)

                    if canonical_smiles not in seen_smiles:
                        seen_smiles.add(canonical_smiles)
                        qed_score = QED.qed(final_mol)
                        mw = Descriptors.ExactMolWt(final_mol)

                        if qed_score > 0.2 and mw < 1200:
                            generated_protacs.append({
                                "ID": f"CRTAP-PROTAC-AI-{i}",
                                "SMILES": canonical_smiles,
                                "QED_Score": round(qed_score, 3),
                                "Molecular_Weight": round(mw, 2),
                                "MolObject": final_mol
                            })
                except:
                    continue

# Step D: Sort and display top 3 results
generated_protacs = sorted(generated_protacs, key=lambda x: x['QED_Score'], reverse=True)
print(f"\nSuccessfully generated and screened {len(generated_protacs)} unique viable candidates!\n")

top_mols = []
top_legends = []

for candidate in generated_protacs[:3]:
    print(f"Candidate: {candidate['ID']}")
    print(f"  Chemical Structure String: {candidate['SMILES']}")
    print(f"  AI Optimization Score (QED): {candidate['QED_Score']}")
    print(f"  Molecular Weight: {candidate['Molecular_Weight']} Da\n")

    top_mols.append(candidate["MolObject"])
    top_legends.append(f"{candidate['ID']} (QED: {candidate['QED_Score']})")

# Step E: Draw the Top 3 Molecules
img = Draw.MolsToGridImage(top_mols, molsPerRow=3, subImgSize=(400, 400), legends=top_legends)
display(img)

# Step F: Export and Download results.csv automatically
export_data = [{k: v for k, v in item.items() if k != 'MolObject'} for item in generated_protacs]
df = pd.DataFrame(export_data)
df.to_csv("results.csv", index=False)
files.download("results.csv")
```
